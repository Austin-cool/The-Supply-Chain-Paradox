```python
# Obtain necessary datasets
import pandas as pd
import matplotlib.pyplot as plt
```

```python
All_stores = pd.read_csv('merged_all_stores.csv')
store_ids = [104, 117, 123, 131, 208, 214, 226, 305, 311, 318]
Departments = ['Health', 'Meat', 'Household', 'Beverages', 'Grocery', 'Frozen', 'Bakery', 'Produce','Dairy']

Data_per_store = {}
for store in store_ids:
    Data_per_store[store] = {}

    Table = All_stores.loc[
            (All_stores["store_id"] == store)
        ]
    
    for department in Departments:
        a = [i for i in Table.columns.tolist() if department in i]
        b = ['date', 'sale_week_type']
        subset = Table[b + a]

        Data_per_store[store][department] = subset
```

```python
for department in Departments:
    dataset = Data_per_store[131][department]
    a = [i for i in dataset.columns.tolist() if "profit" in i]
    b = [i for i in dataset.columns.tolist() if "zero" in i]

    plt.figure(figsize=(10, 5))
    plt.scatter(
        dataset[b[0]],
        dataset[a[0]])
    plt.xlim(-0.05, 1)    
    plt.title(f'Profit in {department} Department')
    plt.xlabel('Weekday')
    plt.ylabel('Profit')
    plt.xticks(rotation=45)
    plt.tight_layout()
    plt.show()
```

```python
import matplotlib.pyplot as plt

fig, axes = plt.subplots(9, 1, figsize=(10, 40), sharex=True)

for ax, department in zip(axes.flat, Departments):
    dataset = Data_per_store[131][department]
    a = [i for i in dataset.columns.tolist() if "profit" in i]
    b = [i for i in dataset.columns.tolist() if "zero" in i]

    ax.scatter(dataset[b[0]], dataset[a[0]])
    ax.set_xlim(-0.05, 1)              # same as your original
    ax.set_title(f'Profit in {department} Department')
    ax.set_ylabel('Profit')
    ax.grid(True, alpha=0.3)
    # y-axis is left to auto-scale, exactly like your original

# Only the bottom subplot needs an x-label (sharex hides the rest)
axes[-1].set_xlabel('Zero rate')

fig.tight_layout()
plt.show()
```
