```python
# Obtain necessary datasets
import pandas as pd
import matplotlib.pyplot as plt
```

```python
All_stores = pd.read_csv('../merged_all_stores.csv')
store_ids = [104, 117, 123, 131, 208, 214, 226, 305, 311, 318]
Departments = ['Health', 'Meat', 'Household', 'Beverages', 'Grocery', 'Frozen', 'Bakery', 'Produce','Dairy']

Data_per_store = {}
for store in store_ids:
    Data_per_store[store] = {}

    Table = All_stores.loc[
            (All_stores["store_id"] == store)
        ]
    
    for department in Departments:
        a = [i for i in Table.columns.tolist() if department in i]
        b = ['date', 'sale_week_type']
        subset = Table[b + a]

        Data_per_store[store][department] = subset
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

for store in store_ids: 
    fig, axes = plt.subplots(9, 1, figsize=(10, 40), sharex=True)
    
    for ax, department in zip(axes.flat, Departments):
        dataset = Data_per_store[store][department]
        a = [i for i in dataset.columns.tolist() if "profit" in i]
        b = [i for i in dataset.columns.tolist() if "zero" in i]
    
        x = dataset[b[0]].to_numpy()
        y = dataset[a[0]].to_numpy()
    
        # Drop NaNs so the regression doesn't choke
        mask = ~(np.isnan(x) | np.isnan(y))
        x, y = x[mask], y[mask]
    
        # Scatter
        ax.scatter(x, y, s=20, alpha=0.6, edgecolor='k', linewidth=0.3, label='data')
    
        # Linear regression
        if len(x) > 1:
            slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
            x_line = np.linspace(0, 1, 100)
            y_line = slope * x_line + intercept
            ax.plot(x_line, y_line, 'r-', linewidth=2,
                    label=f'y = {slope:.2f}x + {intercept:.2f}  (R² = {r_value**2:.3f})')
            ax.legend(loc='best', fontsize=8)
    
        ax.set_xlim(-0.05, 1)
        ax.set_title(f'Profit in {department} Department')
        ax.set_ylabel('Profit')
        ax.grid(True, alpha=0.3)
    
    fig.tight_layout()
    
    # --- Save as SVG ---
    fig.savefig(
        f"Profit vs zero rate per department in {store}.svg",
        bbox_inches='tight'
    )
    
    plt.close(fig)
```

```python
import matplotlib.pyplot as plt
import numpy as np
from scipy import stats

fig, axes = plt.subplots(9, 1, figsize=(10, 40), sharex=True)

for ax, department in zip(axes.flat, Departments):
    dataset = Data_per_store[131][department]
    a = [i for i in dataset.columns.tolist() if "profit" in i]
    b = [i for i in dataset.columns.tolist() if "zero" in i]

    x = dataset['sale_week_type'].to_numpy()
    y = dataset[a[0]].to_numpy()

    # Drop NaNs so the regression doesn't choke
    mask = ~(np.isnan(x) | np.isnan(y))
    x, y = x[mask], y[mask]

    # Scatter
    ax.scatter(x, y, s=20, alpha=0.6, edgecolor='k', linewidth=0.3, label='data')

    # Linear regression
    if len(x) > 1:
        slope, intercept, r_value, p_value, std_err = stats.linregress(x, y)
        x_line = np.linspace(x.min(), x.max(), 100)
        y_line = slope * x_line + intercept
        ax.plot(x_line, y_line, 'r-', linewidth=2,
                label=f'y = {slope:.2f}x + {intercept:.2f}  (R² = {r_value**2:.3f})')
        ax.legend(loc='best', fontsize=8)

    # ax.set_xlim(-0.05, 1)
    ax.set_title(f'Profit in {department} Department')
    ax.set_ylabel('Profit')
    ax.grid(True, alpha=0.3)

fig.tight_layout()
plt.show()
```
